```python
%%writefile app.py
import streamlit as st
import pandas as pd
import numpy as np
import joblib

# ===============================
# Load Model
# ===============================
model = joblib.load("traffic_model.pkl")

st.set_page_config(page_title="Traffic Volume Prediction", page_icon="🚗")

st.title("🚗 Traffic Volume Prediction")
st.write("Enter the weather and date information to predict traffic volume.")

# ===============================
# Date & Time
# ===============================

date = st.date_input("Select Date")
time = st.time_input("Select Time")

dt = pd.to_datetime(f"{date} {time}")

hour = dt.hour
day = dt.day
month = dt.month
year = dt.year

day_num = dt.dayofweek
day_of_week = day_num

is_weekend = 1 if day_num in [5, 6] else 0

rush_hour = 1 if (7 <= hour <= 9 or 16 <= hour <= 19) else 0

is_peak = 1 if hour in [7,8,9,16,17,18,19] else 0

hour_sin = np.sin(2*np.pi*hour/24)
hour_cos = np.cos(2*np.pi*hour/24)

# ===============================
# Weather Inputs
# ===============================

st.subheader("Weather Information")

is_holiday = st.selectbox("Holiday", [0,1])

air_pollution_index = st.number_input(
    "Air Pollution Index",
    min_value=0,
    value=100
)

humidity = st.slider(
    "Humidity",
    0,
    100,
    50
)

wind_speed = st.number_input(
    "Wind Speed",
    min_value=0,
    value=2
)

wind_direction = st.number_input(
    "Wind Direction",
    min_value=0,
    max_value=360,
    value=180
)

temperature = st.number_input(
    "Temperature (Kelvin)",
    value=290.0
)

rain_p_h = st.number_input(
    "Rain per Hour",
    value=0.0
)

snow_p_h = st.number_input(
    "Snow per Hour",
    value=0.0
)

clouds_all = st.slider(
    "Cloud Coverage",
    0,
    100,
    40
)

weather_type = st.selectbox(
    "Weather Type",
    [
        "Clouds",
        "Clear",
        "Rain",
        "Snow",
        "Mist",
        "Drizzle",
        "Haze",
        "Fog",
        "Thunderstorm",
        "Smoke",
        "Squall"
    ]
)

weather_description = st.text_input(
    "Weather Description",
    "scattered clouds"
)

# ===============================
# Prediction
# ===============================

if st.button("Predict Traffic Volume"):

    input_df = pd.DataFrame({

        "is_holiday":[is_holiday],
        "air_pollution_index":[air_pollution_index],
        "humidity":[humidity],
        "wind_speed":[wind_speed],
        "wind_direction":[wind_direction],
        "temperature":[temperature],
        "rain_p_h":[rain_p_h],
        "snow_p_h":[snow_p_h],
        "clouds_all":[clouds_all],
        "weather_type":[weather_type],
        "weather_description":[weather_description],
        "hour":[hour],
        "day_of_week":[day_of_week],
        "day_num":[day_num],
        "month":[month],
        "year":[year],
        "is_weekend":[is_weekend],
        "rush_hour":[rush_hour],
        "day":[day],
        "hour_sin":[hour_sin],
        "hour_cos":[hour_cos],
        "is_peak":[is_peak]

    })

    prediction = model.predict(input_df)

    st.success(
        f"🚦 Predicted Traffic Volume: {prediction[0]:.0f} vehicles"
    )
```
